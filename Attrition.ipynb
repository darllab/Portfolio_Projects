```python
#Attrition Count by Reason for leaving 

attrition.groupby('Reason for Leaving').agg({'Attrition Count': ['count','sum']})
```

```python
#Attrition by Tenure Band

attrition.groupby('Tenure Band').agg({'Attrition Count': ['count','sum']})
```

```python
x = ['0-3 months', '3 months to 6 months']
y = [315,254]
plt.bar(x,y, color = 'orange')
plt.title('Attrition by Tenure Band')
plt.show()
```

```python
#Attrition by Location

attrition.groupby('Location').agg({'Attrition Count': ['sum']})
```

```python
x = ['PHL - Davao', 'PHL - Manila']
y = [315,254]
plt.bar(x,y)
plt.show()
```
